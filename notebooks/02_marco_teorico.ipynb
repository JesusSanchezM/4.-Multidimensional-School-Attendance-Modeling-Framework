```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# FIGURE NUMBER
# =====================================================
FIG_NUM = '2.1'

# =====================================================
# UNIFIED ACADEMIC STYLE
# =====================================================
plt.rcParams.update({
    'font.family': 'serif',
    'font.size': 11,
    'axes.labelsize': 11,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
    'figure.dpi': 300
})

# =====================================================
# FIGURA
# =====================================================
fig, ax = plt.subplots(figsize=(6.5, 4.8))

# Curva sigmoide
x = np.linspace(0, 10, 500)
y = 0.18 + 0.78/(1 + np.exp(-(x - 5)/1.5))
ax.plot(x, y, color='black', linewidth=1.6)

# Línea horizontal sólida (U)
ax.hlines(0.50, 0, 10, colors='black', linewidth=1.2)

# Línea horizontal discontinua (T')
ax.hlines(0.70, 2.2, 9.4,
          colors='black',
          linestyles=(0, (4, 3)),
          linewidth=1.2)

# Escalón discontinuo
ax.vlines(2.2, 0.30, 0.70,
          colors='black',
          linestyles=(0, (4, 3)),
          linewidth=1.2)

ax.hlines(0.30, 0, 2.2,
          colors='black',
          linestyles=(0, (4, 3)),
          linewidth=1.2)

# =====================================================
# ETIQUETAS
# =====================================================
ax.text(-0.22, 0.18, 'T', ha='right', va='center')
ax.text(-0.22, 0.30, "T′", ha='right', va='center')
ax.text(-0.22, 0.50, 'U', ha='right', va='center')

ax.text(10.12, y[-1], 'T', ha='left', va='center')
ax.text(10.12, 0.70, "T′", ha='left', va='center')
ax.text(10.12, 0.50, 'U', ha='left', va='center')

# =====================================================
# FORMATO DE EJES
# =====================================================
ax.set_xlim(0, 10)
ax.set_ylim(0.1, 0.95)

ax.set_xlabel('AGE', fontsize=12)
ax.set_ylabel('EARNINGS', fontsize=12, labelpad=25)

ax.set_xticks([])
ax.set_yticks([])

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.spines['left'].set_linewidth(1.5)
ax.spines['bottom'].set_linewidth(1.5)

# ------------------------------------------------------------------
# 6. GUARDAR EN SVG Y PDF
# ------------------------------------------------------------------
plt.savefig('Figura_2.1.svg', format='svg', bbox_inches='tight')
plt.savefig('Figura_2.1.pdf', format='pdf', bbox_inches='tight')

plt.show()

# ------------------------------------------------------------------
# 7. MOSTRAR TABLA
# ------------------------------------------------------------------
print("\nTabla de países utilizados:")
print("No aplica para esta figura.")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import pycountry
import io

# ------------------------------------------------------------------
# 1. DATOS DE PIB PER CÁPITA (2024)
# ------------------------------------------------------------------
csv_pib = """
Country Code,2024
ABW,38590.56503
AFE,1628.227289
AFG,416.8711463
AFW,1416.228412
AGO,2720.819007
ALB,11374.00858
AND,49357.43707
ARB,7602.411157
ARE,50273.51262
ARG,13969.78366
ARM,8556.21407
ASM,
ATG,23059.83307
AUS,64609.98651
AUT,58268.87877
AZE,7294.642761
BDI,216.2319285
BEL,56581.93756
BEN,1485.380189
BFA,982.4931004
BGD,2593.416117
BGR,17596.91676
BHR,29717.14246
BHS,39455.44665
BIH,9397.909429
BLR,8605.609546
BLZ,7681.244006
BMU,142250.4332
BOL,4421.166099
BRA,10310.5487
BRB,26897.2002
BRN,33153.47389
BTN,4228.051024
BWA,7649.816637
CAF,516.161
CAN,55015.70669
CEB,24616.09579
CHE,107702.0659
CHI,
CHL,16658.96527
CHN,13293.11623
CIV,2727.893522
CMR,1830.008337
COD,692.8876651
COG,2482.248995
COL,7951.08624
COM,1857.870607
CPV,5170.205319
CRI,18853.28287
CSS,19923.57459
CUB,
CUW,22832.8955
CYM,104292.9007
CYP,38674.3125
CZE,31827.75525
DEU,56103.73232
DJI,3552.723351
DMA,10405.27878
DNK,71026.48323
DOM,10875.66184
DZA,5752.990767
EAP,10391.15261
EAR,4405.284926
EAS,13482.47888
ECA,9801.700547
ECS,31630.85529
ECU,6826.529414
EGY,3338.47371
EMU,46424.24131
ERI,
ESP,35326.76831
EST,31428.35478
ETH,1133.882797
EUU,43313.76682
FIN,53155.54512
FJI,6425.741517
FRA,46103.08409
FRO,74175.27763
FSM,4251.690527
GAB,8230.043115
GBR,53341.31311
GEO,8967.675256
GHA,2419.251867
GIB,
GIN,1694.953759
GMB,871.3402875
GNB,998.3760934
GNQ,7003.580557
GRC,24626.14819
GRD,11528.92208
GRL,
GTM,6150.894653
GUM,
GUY,29675.25016
HIC,50907.27615
HKG,54274.75478
HND,3415.960279
HPC,1220.845771
HRV,24049.94417
HTI,2060.355918
HUN,23305.44825
IBD,8186.574761
IBT,6304.397
IDA,1487.280321
IDB,1658.962909
IDN,4925.43645
IDX,1395.67576
IMN,
IND,2591.991661
INX,
IRL,112894.9532
IRN,5190.16959
IRQ,6073.610323
ISL,85863.14793
ISR,54216.52181
ITA,40429.6224
JAM,7753.811953
JOR,5073.921036
JPN,33797.10143
KAZ,14154.6325
KEN,2133.461934
KGZ,2514.810669
KHM,2627.879698
KIR,2550.983772
KNA,23960.65344
KOR,36238.63991
KWT,32855.72097
LAC,10065.50291
LAO,2123.979094
LBN,4473.271344
LBR,851.4977239
LBY,6569.164087
LCA,14493.65847
LCN,10705.10678
LDC,1299.443689
LIC,808.5005646
LIE,220167.2255
LKA,4545.360069
LMC,2354.154176
LMY,5874.030804
LSO,1023.042277
LTE,12874.64364
LTU,29603.77726
LUX,137781.6817
LVA,23578.96636
MAC,72004.7437
MAF,
MAR,4153.194336
MCO,288001.5749
MDA,7578.902163
MDG,550.3787553
MDV,13379.35136
MEA,6275.434696
MEX,13988.04237
MHL,7590.284436
MIC,6507.457272
MKD,9291.856604
MLI,1093.252332
MLT,44023.63411
MMR,1359.050016
MNA,2963.069137
MNE,13270.18244
MNG,6750.630116
MNP,
MOZ,656.9761486
MRT,2104.542608
MUS,11990.93554
MWI,522.3653661
MYS,11874.42737
NAC,82808.1205
NAM,4501.848495
NCL,29213.19232
NER,729.8566642
NGA,1084.160418
NIC,2847.876395
NLD,67465.32081
NOR,89888.95352
NPL,1460.28238
NRU,14048.16396
NZL,49432.36264
OED,48881.81091
OMN,20285.22728
OSS,16111.83282
PAK,1479.477552
PAN,19161.21884
PER,8526.296657
PHL,3985.294647
PLW,18169.07672
PNG,2912.491407
POL,25103.56566
PRE,1372.711535
PRI,39353.22424
PRK,
PRT,29328.30674
PRY,6456.607231
PSE,3028.254813
PSS,4528.483646
PST,56385.05851
PYF,22439.88387
QAT,75685.08658
ROU,20080.20962
RUS,14961.89746
RWA,1059.937093
SAS,2601.822095
SAU,35527.78618
SDN,984.607666
SEN,1738.732238
SGP,94896.56215
SLB,1933.555384
SLE,806.6546503
SLV,5503.103172
SMR,
SOM,629.5388994
SRB,13677.77949
SSA,1541.075734
SSD,
SSF,1542.658443
SST,15409.26486
STP,3502.617681
SUR,6961.789561
SVK,25992.67485
SVN,34301.03187
SWE,57222.71006
SWZ,3909.558924
SXM,41472.58704
SYC,18364.52597
SYR,
TCA,37506.77984
TCD,980.6683121
TEA,10508.24621
TEC,13364.44869
TGO,1266.087777
THA,7386.636151
TJK,1362.0256
TKM,5904.257726
TLA,10441.46858
TLS,1331.970513
TMN,2962.600946
TON,6215.38991
TSA,2601.822095
TSS,1542.658443
TTO,18733.41104
TUN,4187.640794
TUR,15892.71573
TUV,
TZA,1192.606689
UGA,1077.912784
UKR,5391.622559
UMC,10443.34805
URY,24308.4955
USA,86169.66416
UZB,3337.454921
VCT,11501.22652
VEN,4244.457231
VGB,
VIR,
VNM,4716.659938
VUT,3959.877035
WLD,13717.091
WSM,5392.877619
XKX,7026.698813
YEM,
ZAF,6267.186814
ZMB,1187.109434
ZWE,2496.15515
"""

# ------------------------------------------------------------------
# 2. DATOS DE EDUCACIÓN TERCIARIA
# ------------------------------------------------------------------
data_educ = {
    'Country': [
        'Australia','Austria','Belgium','Canada','Colombia','Costa Rica','Czechia',
        'Denmark','Estonia','Finland','France','Germany','Greece','Hungary','Ireland',
        'Israel','Italy','Korea','Latvia','Lithuania','Luxembourg','Mexico',
        'Netherlands','New Zealand','Norway','Poland','Portugal','Slovak Republic',
        'Slovenia','Spain','Sweden','Switzerland','Turkiye','United Kingdom',
        'European Union','OECD','Bulgaria','Croatia','Romania','South Africa'
    ],
    'Tertiary': [
        53.1, 37.7, 45.0, 64.7, 30.6, 27.8, 27.5,
        45.1, 42.5, 42.7, 43.4, 34.3, 35.3, 31.1,
        57.5, 50.5, 22.3, 56.2, 40.5, 47.7, 54.4,
        21.9, 45.1, 44.0, 50.4, 39.5, 31.4, 29.0,
        34.6, 42.3, 51.8, 46.5, 26.9, 53.8,
        38.6, 41.2, 33.8, 30.4, 19.2, 9.0
    ]
}
df_educ = pd.DataFrame(data_educ)

# ------------------------------------------------------------------
# 3. UNIR DATOS
# ------------------------------------------------------------------
df_pib = pd.read_csv(io.StringIO(csv_pib))
df_pib = df_pib.dropna(subset=['2024'])

def code_to_name(code):
    try:
        country = pycountry.countries.get(alpha_3=code)
        return country.name if country else code
    except:
        return code

df_pib['Country_name'] = df_pib['Country Code'].apply(code_to_name)

mapeo_manual = {
    'Czechia': 'Czechia',
    'Korea': 'Korea, Republic of',
    'Slovak Republic': 'Slovakia',
    'United Kingdom': 'United Kingdom of Great Britain and Northern Ireland',
    'Turkiye': 'Turkey',
    'European Union': None,
    'OECD': None,
    'South Africa': 'South Africa'
}
df_educ['Country_clean'] = df_educ['Country'].replace(mapeo_manual)

df_merged = pd.merge(df_educ, df_pib, left_on='Country_clean', right_on='Country_name', how='inner')
df_merged = df_merged.dropna(subset=['Tertiary', '2024'])

print(f"Países usados en la regresión: {len(df_merged)}")

# ------------------------------------------------------------------
# 4. REGRESIÓN
# ------------------------------------------------------------------
x = df_merged['Tertiary'].values
y = df_merged['2024'].values

slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
r2 = r_value ** 2

print(f"\nEcuación: PIB = {slope:.2f} * Tertiary + {intercept:.2f}")
print(f"R² = {r2:.4f}")

x_line = np.linspace(min(x)-2, max(x)+2, 100)
y_line = slope * x_line + intercept

# ------------------------------------------------------------------
# 5. GRÁFICO
# ------------------------------------------------------------------
fig, ax = plt.subplots(figsize=(10, 6))

# Puntos
ax.scatter(x, y, color='steelblue', edgecolors='black', s=80, alpha=0.8)

# Recta de regresión
ax.plot(x_line, y_line, color='crimson', linewidth=2)

# --- Etiquetar los 4 países con mayor PIB y también México ---
top4 = df_merged.nlargest(4, '2024')
paises_a_etiquetar = set(top4['Country'].tolist())
paises_a_etiquetar.add('Mexico')  # Asegurar que México esté siempre etiquetado

for _, row in df_merged.iterrows():
    if row['Country'] in paises_a_etiquetar:
        ax.annotate(row['Country'], (row['Tertiary'], row['2024']),
                    xytext=(5, 5), textcoords='offset points', fontsize=9, fontweight='bold')



ax.set_xlabel('Porcentaje de población con educación terciaria (%)', fontsize=12)
ax.set_ylabel('PIB per cápita (USD)', fontsize=12)
# Sin título (eliminado)
ax.grid(True, linestyle='--', alpha=0.6)
ax.legend(loc='lower right')

plt.tight_layout()

# ------------------------------------------------------------------
# 6. GUARDAR EN SVG Y PDF
# ------------------------------------------------------------------
plt.savefig('Figura_2.2.svg', format='svg', bbox_inches='tight')
plt.savefig('Figura_2.2.pdf', format='pdf', bbox_inches='tight')

plt.show()

# ------------------------------------------------------------------
# 7. MOSTRAR TABLA
# ------------------------------------------------------------------
print("\nTabla de países utilizados:")
print(df_merged[['Country', 'Tertiary', '2024']].to_string(index=False))
```
